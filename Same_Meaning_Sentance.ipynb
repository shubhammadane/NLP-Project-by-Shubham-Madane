Problem Statement

Given two questions:

Question 1:
How can I learn Python quickly?

Question 2:
What is the fastest way to learn Python?

Output

Same Meaning

Another example

Question 1:
How can I learn Python?

Question 2:
What is Java?

Output

Different Meaning

```python
from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim

# Load SBERT model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Questions
question1 = "How can I learn Python quickly?"
question2 = "What is the fastest way to learn Python?"

# Generate embeddings
embedding1 = model.encode(question1)
embedding2 = model.encode(question2)

# Calculate similarity
similarity = cos_sim(embedding1, embedding2).item()

print("Similarity Score:", similarity)

# Decision
threshold = 0.80

if similarity >= threshold:
    print("Same Meaning")
else:
    print("Different Meaning")
```

```python
from sentence_transformers import SentenceTransformer
from sentence_transformers.util import cos_sim

# Load SBERT model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Questions
question1 = "What is Java"
question2 = "What is  Python?"

# Generate embeddings
embedding1 = model.encode(question1)
embedding2 = model.encode(question2)

# Calculate similarity
similarity = cos_sim(embedding1, embedding2).item()

print("Similarity Score:", similarity)

# Decision
threshold = 0.80

if similarity >= threshold:
    print("Same Meaning")
else:
    print("Different Meaning")
```
