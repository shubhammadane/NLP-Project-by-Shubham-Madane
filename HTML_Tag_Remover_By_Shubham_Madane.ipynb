## NLP Mini Project: Removing HTML Tags from Text

This notebook demonstrates how to effectively remove HTML tags from a given text using Python, primarily leveraging the `BeautifulSoup` library. This is a common task in Natural Language Processing (NLP) when raw text data is extracted from web pages or other HTML sources and needs to be cleaned for further analysis.

```python
# NLP Mini Project
# Remove HTML Tags from Text using Python

print("NLP Mini Project - HTML Tag Remover")
```

### 1. Importing Necessary Libraries

We start by importing `BeautifulSoup` from the `bs4` library. `BeautifulSoup` is a Python library for parsing HTML and XML documents. It creates a parse tree for parsed pages that can be used to extract data from HTML, which is perfect for our use case of stripping HTML tags.

```python
from bs4 import BeautifulSoup
```

### 2. Defining Sample HTML Text

Next, we define a multi-line string `html_text` that contains a sample HTML structure. This will be the input text from which we want to remove the HTML tags.

```python
html_text = """
<html>
<head><title>NLP Project</title></head>
<body>
<h1>Welcome to NLP</h1>
<p>This is <b>bold</b> text.</p>
<a href="#">Click Here</a>
</body>
</html>
"""

print(html_text)
```

```python
html_text = """
<html>
<head><title>NLP Project</title></head>
<body>
<h1>Welcome to NLP</h1>
<p>This is <b>bold</b> text.</p>
<a href="#">Click Here</a>
</body>
</html>
"""

# Print the original HTML text to see its structure.
print(html_text)
```

### 3. Cleaning HTML Text (Direct Application)

Here, we directly apply `BeautifulSoup` to parse the `html_text` and then use the `.get_text()` method to extract all the text content, effectively stripping away all HTML tags. The `"html.parser"` argument tells `BeautifulSoup` which parser to use.

```python
clean_text = BeautifulSoup(html_text, "html.parser").get_text()

print(clean_text)
```

```python
# Create a BeautifulSoup object, parsing the HTML text.
# Then, use .get_text() to extract only the visible text content, removing all tags.
clean_text = BeautifulSoup(html_text, "html.parser").get_text()

# Print the cleaned text to observe the result.
print(clean_text)
```

### 4. Encapsulating the Logic in a Function

To make the process reusable, we define a Python function `remove_html_tags` that takes a text string as input, performs the HTML cleaning using `BeautifulSoup`, and returns the cleaned text. This promotes modularity and makes it easy to apply the cleaning to multiple strings.

```python
def remove_html_tags(text):
    clean = BeautifulSoup(text, "html.parser").get_text()
    return clean
```

```python
def remove_html_tags(text):
    # Create a BeautifulSoup object from the input text.
    # The 'html.parser' is used to correctly interpret the HTML structure.
    clean = BeautifulSoup(text, "html.parser").get_text()
    # Return the extracted plain text.
    return clean
```

### 5. Demonstrating the Function

Finally, we demonstrate the usage of our `remove_html_tags` function with another `sample` HTML string. We print both the original sample text and the cleaned result to clearly show the function's effect.

```python
sample = "<p>Hello <b>Shubham</b>, Welcome to <i>NLP</i> Project.</p>"

result = remove_html_tags(sample)

print("Original Text:")
print(sample)

print("\nClean Text:")
print(result)
```
