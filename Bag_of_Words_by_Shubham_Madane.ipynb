```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Create the dataframe
df = pd.DataFrame({'text': ['people watch campus', 'campus watch campus', 'people write comment', 'campus write comment']})

# Display the dataframe
print("DataFrame:")
print(df)
print()
```

### Data Preparation

This section initializes the necessary libraries and creates a sample DataFrame containing text data. This DataFrame will be used to demonstrate the functionality of `CountVectorizer`.

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

### Initializing CountVectorizer

Here, we initialize `CountVectorizer` from `sklearn.feature_extraction.text`. This object will be used to convert a collection of text documents to a matrix of token counts (Bag-of-Words representation).

```python
bow = cv.fit_transform(df['text'])

# vocab
print(cv.vocabulary_)

print(bow[0].toarray())
print(bow[1].toarray())
```

### Fitting and Transforming the Data

We use `cv.fit_transform()` to both learn the vocabulary from the text data in our DataFrame (`df['text']`) and transform it into a Bag-of-Words matrix (`bow`). We then print the learned vocabulary and the Bag-of-Words representation for the first two documents.

```python
# Print the bag-of-words for first two documents
print("Bag-of-Words for document 0 (people watch campus):")
print(bow[0].toarray())
print()

print("Bag-of-Words for document 1 (campus watch campus):")
print(bow[1].toarray())
print()

print("Bag-of-Words for document 0 (people watch campus):")
print(bow[2].toarray())
print()

print("Bag-of-Words for document 1 (campus watch campus):")
print(bow[3].toarray())
print()
```

### Displaying Bag-of-Words for Documents

This section explicitly displays the Bag-of-Words representation for each of the documents in the DataFrame. Each row in the output array corresponds to a document, and the columns represent the count of each word in the vocabulary for that document.

```python
cv.fit(df['text'])
cv.transform(["campusx watch and write comment of campusx"]).toarray()
```

### Transforming New Text

Here, we demonstrate how to transform a new, unseen sentence using the vocabulary learned from the training data. `cv.fit(df['text'])` ensures that the `CountVectorizer` is fitted to the training data first, and then `cv.transform()` converts the new sentence into its Bag-of-Words representation based on that learned vocabulary.
