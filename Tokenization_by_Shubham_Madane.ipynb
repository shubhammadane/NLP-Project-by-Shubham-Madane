What is Tokenization?
=======================
Tokenization means breaking a sentence or paragraph into smaller parts called tokens.

Tokens can be:

1) words

2) sentences

3) characters

Example:
----------

I love NLP

After Tokenization:

['I', 'love', 'NLP']

### Importing Tokenization Functions

Next, we import the specific functions we need from `nltk.tokenize`:
- `word_tokenize`: This function is used for splitting text into individual words.
- `sent_tokenize`: This function is used for splitting text into sentences.

```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
```

### Downloading NLTK Data

NLTK relies on various pre-trained models and data for its functions. The `punkt` tokenizer is crucial for both word and sentence tokenization. We download it once.

*(Note: `punkt_tab` was also downloaded implicitly or explicitly in a previous step, which is part of the `punkt` package and ensures all necessary components are available for different languages.)*

```python
nltk.download('punkt')
```

### Defining Our Sample Text

Let's define a sample paragraph of personal information text that we will use for our tokenization examples. This text will be the input for our word and sentence tokenizers.

```python
text = """
Hello 👋

My name is Shubham Madane.
I am pursuing M.Tech in Computer Science.
I love NLP, Deep Learning and Artificial Intelligence.
Currently I am learning Transformers and LLMs.
"""

print("Original Text:")
print(text)
```

### Performing Word Tokenization

Now, we'll use the `word_tokenize` function to break down our sample text into individual words. This process is very useful for tasks like counting specific words, analyzing vocabulary, or preparing text for further natural language processing.

```python
import nltk
nltk.download('punkt_tab')
word_tokens = word_tokenize(text)

print("Word Tokens:")
print(word_tokens)
```

### Performing Sentence Tokenization

Next, we'll use the `sent_tokenize` function to split our text into complete sentences. This is important for understanding the overall structure and flow of the text, and it's a fundamental step for many NLP tasks like summarization or sentiment analysis at a sentence level.

```python
sentence_tokens = sent_tokenize(text)

print("Sentence Tokens:\n")

for sentence in sentence_tokens:
    print(sentence)
```

### Summarizing Token Counts

Finally, let's see how many word tokens and sentence tokens we extracted from our sample text. These counts give us a quick overview of the text's composition.

```python
print("Total Word Tokens :", len(word_tokens))
print("Total Sentences   :", len(sentence_tokens))
```
