NLP Mini Project
===================
Remove Emoji from Personal Introduction Text
Problem Statement
====================
In Natural Language Processing (NLP), text collected from social media, chats, resumes, and user profiles often contains emojis.
These emojis increase noise in the dataset and affect text preprocessing and model performance.

This project removes emojis from a user's introduction text using Python and Regular Expressions (re library).

### What is this project about?

This project aims to remove emojis (like 😊, 🚀, ❤️) from text data. Emojis are common in social media and chat messages but can sometimes make it harder for computers to understand the text properly. By removing them, we make the text 'cleaner' for further processing in Natural Language Processing (NLP) tasks.

```python
# NLP Mini Project
# Emoji Removal from Text

print("NLP Project : Remove Emoji from Text")
```

```python
import re
```

### Step 1: Import necessary tools

We need a special tool called `re` (which stands for Regular Expressions) in Python. This tool helps us find and replace specific patterns in text, which is perfect for finding and removing emojis.

```python
text = """
Hello 😊
My name is Shubham Madane🚀
I am pursuing M.Tech in Computer Science ❤️
I love NLP, Deep Learning and AI 🤖🔥
"""

print("Original Text:")
print(text)
```

### Step 2: Define our text with emojis

Here, we create a sample text that contains several emojis. This `text` variable will be used to demonstrate how our emoji removal function works.

```python
def remove_emoji(text):

    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map
        u"\U0001F700-\U0001F77F"
        u"\U0001F780-\U0001F7FF"
        u"\U0001F800-\U0001F8FF"
        u"\U0001F900-\U0001F9FF"  # supplemental symbols
        u"\U0001FA00-\U0001FAFF"
        u"\U00002702-\U000027B0"
        u"\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )

    return emoji_pattern.sub(r'', text)
```

### Step 3: Create a function to remove emojis

This is the core part of our project. We define a function called `remove_emoji` that takes any text as input. Inside this function:

*   We create an `emoji_pattern` using `re.compile()`. This pattern is a special code that matches a wide range of Unicode emojis.
*   `re.UNICODE` flag makes sure the pattern works correctly with all sorts of international characters, including emojis.
*   Finally, `emoji_pattern.sub(r'', text)` replaces all found emojis with an empty string (`''`), effectively removing them from the text. The `r''` ensures no special characters are misinterpreted.

```python
clean_text = remove_emoji(text)

print("Clean Text:")
print(clean_text)
```

### Step 4: Apply the function and see the result

We call our `remove_emoji` function with our `text` variable and store the output in `clean_text`. Then, we print both the original and the cleaned text to see the difference.

```python
print("========== BEFORE ==========")
print(text)

print("\n========== AFTER ==========")
print(clean_text)
```

### Step 5: Verify the cleaning

This section just prints the original and the cleaned text side-by-side to make it easy to compare and see that the emojis have indeed been removed.

```python
examples = [
    "I love AI ❤️",
    "Python is amazing 🚀",
    "NLP is future 🤖🔥",
    "Machine Learning 😎"
]

for sentence in examples:

    cleaned = remove_emoji(sentence)

    print("Original :", sentence)
    print("Cleaned  :", cleaned)
    print("-" * 40)
```
